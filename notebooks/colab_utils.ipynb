```python
# Слайдер-«листалка» через ipywidgets
# Это даёт тот же UX «листать вперёд/назад», только в ноутбуке.

import json
from pathlib import Path
from IPython.display import display, Image
import ipywidgets as widgets

SAMPLE_DIR = Path("data/samples/processed")
meta = json.load(open(SAMPLE_DIR / "metadata.json"))
paths = [m["output"] for m in meta]

slider = widgets.IntSlider(min=0, max=len(paths) - 1, step=1, description="Кадр")
out = widgets.Output()

def render(idx):
    with out:
        out.clear_output(wait=True)
        display(Image(filename=paths[idx], width=800))

widgets.interactive_output(render, {"idx": slider})
display(slider, out)
```
